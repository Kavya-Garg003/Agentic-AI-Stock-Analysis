````python
import os
import json
import requests
from flask import Flask, request, jsonify, render_template
from transformers import pipeline
import yfinance as yf
import feedparser
import pandas as pd
import numpy as np
from urllib.parse import quote_plus
from datetime import datetime, timedelta
import warnings

warnings.filterwarnings('ignore')

# Initialize Flask App
app = Flask(__name__)

# -------------------------
# Configuration
# -------------------------

OPENROUTER_API_KEY = "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"  # Replace with your actual API key
OPENROUTER_BASE_URL = "https://openrouter.ai/api/v1/chat/completions"

# -------------------------
# Helper: Generate fallback 3-month price data (~63 trading days)
# -------------------------
def generate_fallback_price_data(days=63):
    end_date = datetime.today()
    dates = [end_date - timedelta(days=x) for x in range(days)]
    dates.reverse()  # oldest first

    # Generate some realistic price movement
    base_price = 100
    prices = []
    for i in range(days):
        # Add some random walk with slight upward bias
        change = np.random.normal(0.002, 0.02)  # 0.2% daily drift, 2% volatility
        if i == 0:
            prices.append(base_price)
        else:
            prices.append(prices[-1] * (1 + change))
    
    data = {
        'Open': prices,
        'High': [p * (1 + abs(np.random.normal(0, 0.01))) for p in prices],
        'Low': [p * (1 - abs(np.random.normal(0, 0.01))) for p in prices],
        'Close': prices,
        'Volume': [int(np.random.uniform(100000, 1000000)) for _ in range(days)],
    }

    df = pd.DataFrame(data, index=pd.to_datetime(dates))
    return df

# -------------------------
# OpenRouter API for CANSLIM Analysis
# -------------------------
class CANSLIMAnalyzer:
    def __init__(self, api_key):
        self.api_key = api_key
        self.base_url = OPENROUTER_BASE_URL
        
    def analyze_canslim(self, company, ticker, financials, price_history, market_history, news_headlines):
        """Use DeepSeek via OpenRouter to perform comprehensive CANSLIM analysis"""
        
        # Prepare data for the AI model
        financial_summary = self._prepare_financial_summary(financials, price_history, market_history)
        news_summary = self._prepare_news_summary(news_headlines)
        
        prompt = f"""
You are an expert financial analyst specializing in CANSLIM methodology created by William O'Neil. 
Analyze {company} ({ticker}) and provide a comprehensive CANSLIM score and recommendation.

CANSLIM CRITERIA (Score each 0-1, total 7):
C - Current Earnings: Look for 25%+ quarterly earnings growth
A - Annual Earnings: 25%+ annual earnings growth over 3 years
N - New (Something New): New products, services, management, or high stock price
S - Supply & Demand: Share volume increase, stock buybacks, institutional buying
L - Leader or Laggard: Stock outperforming market and industry
I - Institutional Sponsorship: Quality institutional ownership increasing
M - Market Direction: Overall market trend (3 of 4 recent quarters up)

FINANCIAL DATA:
{financial_summary}

NEWS & MARKET DATA:
{news_summary}

ADDITIONAL FINANCIAL RATIOS TO CONSIDER:
- P/E Ratio vs industry average
- ROE (Return on Equity) - should be >15%
- Debt-to-Equity ratio - lower is better
- Price-to-Book ratio
- Revenue growth consistency
- Free cash flow trends
- Institutional ownership percentage

PROVIDE:
1. Detailed analysis for each CANSLIM letter (C,A,N,S,L,I,M)
2. Score for each letter (0 or 1)
3. Overall CANSLIM score (0-7)
4. BUY/HOLD/SELL recommendation with reasoning
5. Risk assessment (LOW/MEDIUM/HIGH)
6. Key strengths and weaknesses

Format your response as JSON:
{{
    "canslim_analysis": {{
        "C": {{"score": 0/1, "reasoning": "detailed explanation"}},
        "A": {{"score": 0/1, "reasoning": "detailed explanation"}},
        "N": {{"score": 0/1, "reasoning": "detailed explanation"}},
        "S": {{"score": 0/1, "reasoning": "detailed explanation"}},
        "L": {{"score": 0/1, "reasoning": "detailed explanation"}},
        "I": {{"score": 0/1, "reasoning": "detailed explanation"}},
        "M": {{"score": 0/1, "reasoning": "detailed explanation"}}
    }},
    "total_score": 0-7,
    "recommendation": "BUY/HOLD/SELL",
    "risk_level": "LOW/MEDIUM/HIGH",
    "confidence": 0.0-1.0,
    "key_strengths": ["strength1", "strength2"],
    "key_weaknesses": ["weakness1", "weakness2"],
    "summary": "Brief summary of analysis and reasoning"
}}
"""

        try:
            headers = {
                "Authorization": f"Bearer {self.api_key}",
                "Content-Type": "application/json",
                "HTTP-Referer": "https://github.com/your-repo",  # Optional: Your site URL
                "X-Title": "Stock Analysis CANSLIM Tool",  # Optional: Your site name
            }
            
            payload = {
                "model": "deepseek/deepseek-chat",
                "messages": [
                    {"role": "system", "content": "You are an expert financial analyst specializing in CANSLIM methodology. Always respond with valid JSON format."},
                    {"role": "user", "content": prompt}
                ],
                "temperature": 0.3, "max_tokens": 2000, "top_p": 1,
                "frequency_penalty": 0, "presence_penalty": 0
            }
            
            print(f"Making API request to OpenRouter for {company}...")
            response = requests.post(self.base_url, headers=headers, json=payload, timeout=60)
            print(f"Response status code: {response.status_code}")
            
            if response.status_code != 200:
                print(f"API Error Response: {response.text}")
                response.raise_for_status()
            
            result = response.json()
            
            if 'choices' not in result or len(result['choices']) == 0:
                print(f"Unexpected API response structure: {result}")
                raise ValueError("Invalid response structure from OpenRouter API")
            
            content = result['choices'][0]['message']['content']
            print(f"Raw API response: {content[:200]}...")
            
            try:
                if "```json" in content:
                    json_start = content.find("```json") + 7
                    json_end = content.rfind("```")
                    content = content[json_start:json_end].strip()
                elif "```" in content:
                    json_start = content.find("```") + 3
                    json_end = content.rfind("```")
                    content = content[json_start:json_end].strip()
                
                analysis = json.loads(content)
                print("Successfully parsed CANSLIM analysis from API")
                return analysis
            except json.JSONDecodeError as e:
                print(f"JSON parsing failed: {e}")
                print(f"Content to parse: {content}")
                return self._fallback_canslim_analysis(financials, price_history, market_history, news_headlines)
                
        except requests.exceptions.RequestException as e:
            print(f"OpenRouter API request failed: {e}")
            return self._fallback_canslim_analysis(financials, price_history, market_history, news_headlines)
        except Exception as e:
            print(f"OpenRouter API failed with unexpected error: {e}")
            return self._fallback_canslim_analysis(financials, price_history, market_history, news_headlines)
    
    def _prepare_financial_summary(self, financials, price_history, market_history):
        if not financials: return "No financial data available"
        try:
            current_price = price_history['Close'].iloc[-1] if not price_history.empty else 100
            price_change_3m = ((current_price - price_history['Close'].iloc[0]) / price_history['Close'].iloc[0] * 100) if len(price_history) > 1 else 0
            market_change_3m = 0
            if not market_history.empty and len(market_history) > 1:
                market_change_3m = ((market_history['Close'].iloc[-1] - market_history['Close'].iloc[0]) / market_history['Close'].iloc[0] * 100)
            
            summary = f"""
Current Price: ${financials.get('currentPrice', 'N/A')}
P/E Ratio: {financials.get('trailingPE', 'N/A')}
EPS: ${financials.get('trailingEps', 'N/A')}
ROE: {financials.get('returnOnEquity', 'N/A')}%
Debt-to-Equity: {financials.get('debtToEquity', 'N/A')}
Quick Ratio: {financials.get('quickRatio', 'N/A')}
Beta: {financials.get('beta', 'N/A')}
Earnings Growth (Quarterly): {financials.get('earningsQuarterlyGrowth', 'N/A')}%
Earnings Growth (Annual): {financials.get('earningsGrowth', 'N/A')}%
Shares Outstanding: {financials.get('sharesOutstanding', 'N/A')}
Institutional Ownership: {financials.get('heldPercentInstitutions', 'N/A')}%
Volume: {financials.get('volume', 'N/A')}
Stock Performance (3M): {price_change_3m:.2f}%
Market Performance (3M): {market_change_3m:.2f}%
            """
            return summary
        except Exception as e: return f"Error preparing financial summary: {e}"
    
    def _prepare_news_summary(self, news_headlines):
        if not news_headlines or news_headlines == ["No news available."]: return "No recent news available"
        return "Recent Headlines:\n" + "\n".join([f"- {headline}" for headline in news_headlines[:10]])
    
    def _fallback_canslim_analysis(self, financials, price_history, market_history, news_headlines):
        print("Using fallback CANSLIM analysis...")
        financials = financials or {}
        canslim_scores = {}
        
        quarterly_growth = financials.get('earningsQuarterlyGrowth', 0) or 0
        c_score = 1 if quarterly_growth > 0.25 else 0
        canslim_scores['C'] = {"score": c_score, "reasoning": f"Quarterly earnings growth: {quarterly_growth*100:.1f}% ({'✓' if c_score else '✗'} >25% threshold)"}
        
        annual_growth = financials.get('earningsGrowth', 0) or 0
        a_score = 1 if annual_growth > 0.25 else 0
        canslim_scores['A'] = {"score": a_score, "reasoning": f"Annual earnings growth: {annual_growth*100:.1f}% ({'✓' if a_score else '✗'} >25% threshold)"}
        
        n_keywords = ['new product', 'launch', 'acquisition', 'partnership', 'CEO', 'innovation', 'contract', 'breakthrough', 'expansion']
        n_score, new_factors = 0, []
        for headline in (news_headlines or []):
            if any(keyword.lower() in headline.lower() for keyword in n_keywords): n_score=1; break
        canslim_scores['N'] = {"score": n_score, "reasoning": f"New catalysts mentioned in news ({'✓' if n_score else '✗'})"}
        
        try:
            recent_vol = price_history['Volume'][-10:].mean() if len(price_history) > 10 else 0
            earlier_vol = price_history['Volume'][:-10].mean() if len(price_history) > 20 else recent_vol
            s_score = 1 if recent_vol > earlier_vol * 1.5 else 0
        except: s_score = 0
        canslim_scores['S'] = {"score": s_score, "reasoning": f"Volume analysis: {'Increasing' if s_score else 'Stable/Decreasing'} ({'✓' if s_score else '✗'})"}
        
        try:
            stock_return = (price_history['Close'].iloc[-1] - price_history['Close'].iloc[0]) / price_history['Close'].iloc[0]
            market_return = (market_history['Close'].iloc[-1] - market_history['Close'].iloc[0]) / market_history['Close'].iloc[0] if len(market_history) > 1 else 0
            l_score = 1 if stock_return > market_return else 0
        except: l_score = 0
        canslim_scores['L'] = {"score": l_score, "reasoning": f"Performance vs market: {'Outperforming' if l_score else 'Underperforming'} ({'✓' if l_score else '✗'})"}
        
        inst_ownership = financials.get('heldPercentInstitutions', 0) or 0
        i_score = 1 if inst_ownership > 0.3 else 0
        canslim_scores['I'] = {"score": i_score, "reasoning": f"Institutional ownership: {inst_ownership*100:.1f}% ({'✓' if i_score else '✗'} >30% threshold)"}
        
        try:
            market_return = (market_history['Close'].iloc[-1] - market_history['Close'].iloc[0]) / market_history['Close'].iloc[0] if len(market_history) > 1 else 0
            m_score = 1 if market_return > 0 else 0
        except: m_score = 0
        canslim_scores['M'] = {"score": m_score, "reasoning": f"Market direction: {'Positive' if m_score else 'Negative'} trend ({'✓' if m_score else '✗'})"}
        
        total_score = sum(item['score'] for item in canslim_scores.values())
        if total_score >= 5: recommendation, risk_level = "BUY", "LOW" if total_score >= 6 else "MEDIUM"
        elif total_score >= 3: recommendation, risk_level = "HOLD", "MEDIUM"
        else: recommendation, risk_level = "SELL", "HIGH"
        
        return {
            "canslim_analysis": canslim_scores, "total_score": total_score, "recommendation": recommendation, "risk_level": risk_level,
            "confidence": min(0.9, total_score / 7 * 0.8 + 0.2),
            "key_strengths": [k for k, v in canslim_scores.items() if v['score'] == 1],
            "key_weaknesses": [k for k, v in canslim_scores.items() if v['score'] == 0],
            "summary": f"Fallback CANSLIM Score: {total_score}/7. {recommendation} recommendation based on rules."
        }

# -------------------------
# 1. Enhanced Data Collector Agent
# -------------------------
class DataCollectorAgent:
    def __init__(self, company, ticker):
        self.company = company
        self.ticker = ticker
        self.report_folder = "stock_reports"
        os.makedirs(self.report_folder, exist_ok=True)

    def fetch_news(self):
        try:
            print("Attempting to fetch news from API...")
            query = quote_plus(self.company)
            feed_url = f"https://news.google.com/rss/search?q={query}"
            feed = feedparser.parse(feed_url)
            headlines = [entry.title for entry in feed.entries[:15]]
            if not headlines: raise ValueError("No headlines returned from API")
            print("Successfully fetched news from API.")
            return headlines
        except Exception as e:
            print(f"API failed for news: {e}. Trying to load from local CSV...")
            filepath = f"{self.report_folder}/{self.ticker}_sentiment.csv"
            try:
                df = pd.read_csv(filepath)
                print(f"Successfully loaded headlines from {filepath}")
                return df['Headline'].tolist()
            except FileNotFoundError:
                print(f"Local file not found: {filepath}. No news available.")
                return ["No news available."]

    def fetch_financials(self):
        try:
            print("Attempting to fetch financials from API...")
            stock = yf.Ticker(self.ticker)
            info = stock.info
            financials = {
                'currentPrice': info.get('currentPrice'), 'trailingPE': info.get('trailingPE'),
                'forwardPE': info.get('forwardPE'), 'priceToBook': info.get('priceToBook'),
                'priceToSalesTrailing12Months': info.get('priceToSalesTrailing12Months'),
                'enterpriseToRevenue': info.get('enterpriseToRevenue'), 'enterpriseToEbitda': info.get('enterpriseToEbitda'),
                'trailingEps': info.get('trailingEps'), 'forwardEps': info.get('forwardEps'),
                'returnOnEquity': info.get('returnOnEquity'), 'returnOnAssets': info.get('returnOnAssets'),
                'grossMargins': info.get('grossMargins'), 'operatingMargins': info.get('operatingMargins'),
                'profitMargins': info.get('profitMargins'), 'debtToEquity': info.get('debtToEquity'),
                'currentRatio': info.get('currentRatio'), 'quickRatio': info.get('quickRatio'),
                'beta': info.get('beta'), 'earningsQuarterlyGrowth': info.get('earningsQuarterlyGrowth'),
                'earningsGrowth': info.get('earningsGrowth'), 'revenueGrowth': info.get('revenueGrowth'),
                'sharesOutstanding': info.get('sharesOutstanding'), 'heldPercentInstitutions': info.get('heldPercentInstitutions'),
                'volume': info.get('volume'), 'averageVolume': info.get('averageVolume'),
                'marketCap': info.get('marketCap'), 'enterpriseValue': info.get('enterpriseValue'),
                'freeCashflow': info.get('freeCashflow'), 'totalCash': info.get('totalCash'),
                'totalDebt': info.get('totalDebt'), 'dividendYield': info.get('dividendYield'),
                'payoutRatio': info.get('payoutRatio'),
            }
            if financials['currentPrice'] is None: raise ValueError("Incomplete data from yfinance API")
            print("Successfully fetched financials from API.")
            return financials
        except Exception as e:
            print(f"API failed for financials: {e}. Trying to load from local CSV...")
            filepath = f"{self.report_folder}/{self.ticker}_financials.csv"
            try:
                df = pd.read_csv(filepath)
                print(f"Successfully loaded financials from {filepath}")
                financials_dict = df.iloc[0].to_dict()
                return {k: (None if pd.isna(v) else v) for k, v in financials_dict.items()}
            except FileNotFoundError:
                print(f"Local file not found: {filepath}. Using hardcoded fallback data.")
                return {
                    'currentPrice': 100, 'trailingPE': 15, 'forwardPE': 14, 'priceToBook': 2.5,
                    'trailingEps': 6.67, 'returnOnEquity': 0.15, 'returnOnAssets': 0.08,
                    'grossMargins': 0.25, 'operatingMargins': 0.15, 'profitMargins': 0.10,
                    'debtToEquity': 50, 'currentRatio': 1.5, 'quickRatio': 1.2, 'beta': 1.1,
                    'earningsQuarterlyGrowth': 0.10, 'earningsGrowth': 0.12, 'revenueGrowth': 0.08,
                    'sharesOutstanding': 1000000, 'heldPercentInstitutions': 0.65, 'volume': 1000000,
                    'marketCap': 100000000, 'freeCashflow': 10000000, 'totalCash': 5000000,
                    'totalDebt': 2500000, 'dividendYield': 0.02, 'payoutRatio': 0.3,
                }

    def fetch_price_history(self, period="3mo"):
        try:
            print("Attempting to fetch price history from API...")
            stock = yf.Ticker(self.ticker)
            df = stock.history(period=period)
            if df.empty: raise ValueError("Empty DataFrame returned from API")
            print("Successfully fetched price history from API.")
            return df
        except Exception as e:
            print(f"API failed for price history: {e}. Trying to load from local CSV...")
            filepath = f"{self.report_folder}/{self.ticker}_prices.csv"
            try:
                df = pd.read_csv(filepath, index_col=0, parse_dates=True)
                print(f"Successfully loaded price history from {filepath}")
                return df
            except FileNotFoundError:
                print(f"Local file not found: {filepath}. Using generated fallback data.")
                return generate_fallback_price_data()

    def fetch_market_index_history(self, index_ticker="^GSPC", period="3mo"):
        try:
            index = yf.Ticker(index_ticker)
            df = index.history(period=period)
            if df.empty: return generate_fallback_price_data(days=63)
            return df
        except:
            return generate_fallback_price_data(days=63)

# -------------------------------
# 2. Enhanced Fundamental Analyst Agent
# -------------------------------
class FundamentalAnalystAgent:
    def __init__(self, company, data_agent):
        self.company = company
        self.data_agent = data_agent

    def comprehensive_analysis(self, financials):
        analysis = {}
        financials = financials or {}
        analysis['valuation'] = {
            'pe_ratio': financials.get('trailingPE'), 'pb_ratio': financials.get('priceToBook'),
            'ps_ratio': financials.get('priceToSalesTrailing12Months'), 'ev_ebitda': financials.get('enterpriseToEbitda'),
        }
        analysis['profitability'] = {
            'gross_margin': financials.get('grossMargins'), 'operating_margin': financials.get('operatingMargins'),
            'profit_margin': financials.get('profitMargins'), 'roe': financials.get('returnOnEquity'),
            'roa': financials.get('returnOnAssets'),
        }
        analysis['liquidity'] = {
            'current_ratio': financials.get('currentRatio'), 'quick_ratio': financials.get('quickRatio'),
        }
        analysis['leverage'] = {
            'debt_to_equity': financials.get('debtToEquity'), 'total_debt': financials.get('totalDebt'),
            'total_cash': financials.get('totalCash'),
        }
        analysis['growth'] = {
            'earnings_quarterly_growth': financials.get('earningsQuarterlyGrowth'),
            'earnings_growth': financials.get('earningsGrowth'), 'revenue_growth': financials.get('revenueGrowth'),
        }
        analysis['market'] = {
            'market_cap': financials.get('marketCap'), 'beta': financials.get('beta'),
            'institutional_ownership': financials.get('heldPercentInstitutions'),
            'volume': financials.get('volume'), 'average_volume': financials.get('averageVolume'),
        }
        return analysis

# -----------------------------
# 3. Enhanced Sentiment Analyzer Agent
# -----------------------------
class SentimentAnalyzerAgent:
    def __init__(self):
        try:
            self.model = pipeline("sentiment-analysis", model="ProsusAI/finbert")
            print("FinBERT model loaded successfully.")
        except Exception as e:
            print(f"Could not load FinBERT model: {e}. Sentiment analysis will be neutral.")
            self.model = None

    def analyze(self, headlines):
        if not self.model or not headlines or headlines == ["No news available."]:
            return 0, [{'label': 'NEUTRAL', 'score': 1.0} for _ in (headlines or [""])]
        try:
            sentiments = self.model(headlines)
            score_map = {'POSITIVE': 1, 'NEUTRAL': 0, 'NEGATIVE': -1}
            weighted_score, total_weight = 0, 0
            for sentiment in sentiments:
                confidence = sentiment['score']
                if confidence > 0.7:
                    weight = confidence
                    score = score_map.get(sentiment['label'].upper(), 0)
                    weighted_score += score * weight
                    total_weight += weight
            final_score = weighted_score / total_weight if total_weight > 0 else 0
            return final_score, sentiments
        except Exception as e:
            print(f"Error during sentiment analysis: {e}. Defaulting to neutral.")
            return 0, [{'label': 'NEUTRAL', 'score': 1.0} for _ in headlines]

# -------------------------
# 4. Enhanced Risk Manager Agent
# -------------------------
class RiskManagerAgent:
    def assess(self, financials, price_history):
        financials = financials or {}
        risk_factors, risk_score, max_score = {}, 0, 10
        debt_equity = financials.get('debtToEquity', 100)
        if debt_equity is not None:
            if debt_equity < 50: risk_score += 1; risk_factors['debt_level'] = 'Low'
            elif debt_equity < 100: risk_factors['debt_level'] = 'Moderate'
            else: risk_factors['debt_level'] = 'High'
        current_ratio = financials.get('currentRatio', 1.0)
        if current_ratio is not None:
            if current_ratio > 1.5: risk_score += 1; risk_factors['liquidity'] = 'Good'
            elif current_ratio > 1.0: risk_factors['liquidity'] = 'Adequate'
            else: risk_factors['liquidity'] = 'Poor'
        quick_ratio = financials.get('quickRatio', 1.0)
        if quick_ratio is not None and quick_ratio > 1.0: risk_score += 1; risk_factors['quick_liquidity'] = 'Good'
        else: risk_factors['quick_liquidity'] = 'Concern'
        beta = financials.get('beta', 1.5)
        if beta is not None:
            if beta < 1.2: risk_score += 1; risk_factors['market_risk'] = 'Low'
            elif beta < 1.5: risk_factors['market_risk'] = 'Moderate'
            else: risk_factors['market_risk'] = 'High'
        roe = financials.get('returnOnEquity', 0)
        if roe is not None:
            if roe > 0.15: risk_score += 1; risk_factors['profitability'] = 'Strong'
            elif roe > 0.10: risk_factors['profitability'] = 'Adequate'
            else: risk_factors['profitability'] = 'Weak'
        profit_margin = financials.get('profitMargins', 0)
        if profit_margin is not None:
            if profit_margin > 0.15: risk_score += 1; risk_factors['margins'] = 'Healthy'
            elif profit_margin > 0.05: risk_factors['margins'] = 'Acceptable'
            else: risk_factors['margins'] = 'Concerning'
        try:
            if not price_history.empty and len(price_history) > 30:
                volatility = price_history['Close'].pct_change().dropna().std() * np.sqrt(252)
                if volatility < 0.20: risk_score += 1; risk_factors['volatility'] = 'Low'
                elif volatility < 0.35: risk_factors['volatility'] = 'Moderate'
                else: risk_factors['volatility'] = 'High'
            else: risk_factors['volatility'] = 'Unknown'
        except: risk_factors['volatility'] = 'Unknown'
        earnings_growth = financials.get('earningsGrowth', 0)
        if earnings_growth is not None and earnings_growth > 0.10: risk_score += 1; risk_factors['growth_sustainability'] = 'Positive'
        else: risk_factors['growth_sustainability'] = 'Questionable'
        return risk_score, risk_factors, max_score

# ---------------------------
# 5. Enhanced Decision Engine Agent
# ---------------------------
class DecisionEngineAgent:
    def decide(self, fundamental_analysis, sentiment_score, risk_assessment, canslim_analysis=None):
        total_score, max_score, decision_factors = 0, 100, {}
        fundamental_score = self._score_fundamentals(fundamental_analysis)
        total_score += fundamental_score
        decision_factors['fundamental_score'] = fundamental_score
        if canslim_analysis:
            canslim_score = canslim_analysis.get('total_score', 0)
            canslim_weighted = (canslim_score / 7) * 30
            total_score += canslim_weighted
            decision_factors['canslim_score'] = canslim_weighted
        else: decision_factors['canslim_score'] = 0
        sentiment_weighted = max(0, min(15, (sentiment_score + 1) * 7.5))
        total_score += sentiment_weighted
        decision_factors['sentiment_score'] = sentiment_weighted
        risk_score, _, max_risk = risk_assessment
        risk_weighted = (risk_score / max_risk) * 15
        total_score += risk_weighted
        decision_factors['risk_score'] = risk_weighted
        if total_score >= 70: recommendation, confidence = "BUY", min(0.95, total_score / 100)
        elif total_score >= 50: recommendation, confidence = "HOLD", min(0.85, total_score / 100)
        else: recommendation, confidence = "SELL", max(0.60, total_score / 100)
        return {'recommendation': recommendation, 'total_score': round(total_score, 2),
                'max_score': max_score, 'confidence': round(confidence, 3),
                'decision_factors': decision_factors,
                'reasoning': self._generate_reasoning(fundamental_analysis, sentiment_score, risk_assessment, canslim_analysis)}
    
    def _score_fundamentals(self, analysis):
        score = 0
        pe = analysis.get('valuation', {}).get('pe_ratio'); pb = analysis.get('valuation', {}).get('pb_ratio')
        ev_ebitda = analysis.get('valuation', {}).get('ev_ebitda'); ps = analysis.get('valuation', {}).get('ps_ratio')
        roe = analysis.get('profitability', {}).get('roe'); profit_margin = analysis.get('profitability', {}).get('profit_margin')
        gross_margin = analysis.get('profitability', {}).get('gross_margin'); op_margin = analysis.get('profitability', {}).get('operating_margin')
        current_ratio = analysis.get('liquidity', {}).get('current_ratio'); quick_ratio = analysis.get('liquidity', {}).get('quick_ratio')
        earn_growth = analysis.get('growth', {}).get('earnings_growth'); rev_growth = analysis.get('growth', {}).get('revenue_growth')
        q_growth = analysis.get('growth', {}).get('earnings_quarterly_growth')
        if pe is not None:
            if 10 <= pe <= 20: score += 3
            elif 5 <= pe <= 25: score += 2
            elif pe > 0: score += 1
        if pb is not None:
            if 1 <= pb <= 3: score += 2
            elif pb > 0: score += 1
        if ev_ebitda is not None:
            if 5 <= ev_ebitda <= 15: score += 3
            elif ev_ebitda > 0: score += 1
        if ps is not None and ps < 5: score += 2
        if roe is not None:
            if roe > 0.20: score += 5
            elif roe > 0.15: score += 4
            elif roe > 0.10: score += 3
            elif roe > 0.05: score += 1
        if profit_margin is not None:
            if profit_margin > 0.20: score += 4
            elif profit_margin > 0.10: score += 3
            elif profit_margin > 0.05: score += 2
            elif profit_margin > 0: score += 1
        if gross_margin is not None:
            if gross_margin > 0.40: score += 3
            elif gross_margin > 0.25: score += 2
            elif gross_margin > 0.15: score += 1
        if op_margin is not None:
            if op_margin > 0.15: score += 3
            elif op_margin > 0.10: score += 2
            elif op_margin > 0.05: score += 1
        if current_ratio is not None:
            if current_ratio > 2: score += 3
            elif current_ratio > 1.5: score += 2
            elif current_ratio > 1: score += 1
        if quick_ratio is not None and quick_ratio > 1: score += 2
        if earn_growth is not None:
            if earn_growth > 0.25: score += 4
            elif earn_growth > 0.15: score += 3
            elif earn_growth > 0.05: score += 2
            elif earn_growth > 0: score += 1
        if rev_growth is not None:
            if rev_growth > 0.20: score += 3
            elif rev_growth > 0.10: score += 2
            elif rev_growth > 0.05: score += 1
        if q_growth is not None:
            if q_growth > 0.25: score += 3
            elif q_growth > 0.15: score += 2
            elif q_growth > 0: score += 1
        return min(40, score)
    
    def _generate_reasoning(self, fundamental_analysis, sentiment_score, risk_assessment, canslim_analysis):
        reasoning = []
        pe = fundamental_analysis.get('valuation', {}).get('pe_ratio')
        roe = fundamental_analysis.get('profitability', {}).get('roe')
        earn_growth = fundamental_analysis.get('growth', {}).get('earnings_growth')
        if pe:
            if pe < 15: reasoning.append(f"Attractive P/E ratio of {pe:.1f}")
            elif pe > 25: reasoning.append(f"High P/E ratio of {pe:.1f} suggests premium valuation")
        if roe:
            if roe > 0.15: reasoning.append(f"Strong ROE of {roe*100:.1f}%")
            elif roe < 0.10: reasoning.append(f"Low ROE of {roe*100:.1f}% indicates poor capital efficiency")
        if earn_growth:
            if earn_growth > 0.15: reasoning.append(f"Strong earnings growth of {earn_growth*100:.1f}%")
            elif earn_growth < 0: reasoning.append(f"Declining earnings growth of {earn_growth*100:.1f}%")
        if canslim_analysis:
            canslim_score = canslim_analysis.get('total_score', 0)
            if canslim_score >= 5: reasoning.append(f"Strong CANSLIM score of {canslim_score}/7")
            elif canslim_score <= 2: reasoning.append(f"Weak CANSLIM score of {canslim_score}/7")
        if sentiment_score > 0.3: reasoning.append("Positive market sentiment")
        elif sentiment_score < -0.3: reasoning.append("Negative market sentiment")
        risk_score, _, _ = risk_assessment
        if risk_score >= 7: reasoning.append("Low risk profile")
        elif risk_score <= 3: reasoning.append("High risk profile")
        return "; ".join(reasoning) if reasoning else "Analysis based on available financial metrics"

# -------------------------
# 6. Data Persistence Manager
# -------------------------
class DataPersistenceManager:
    def __init__(self, ticker):
        self.ticker = ticker
        self.report_folder = "stock_reports"
        os.makedirs(self.report_folder, exist_ok=True)
    def save_financials(self, financials):
        try:
            pd.DataFrame([financials]).to_csv(f"{self.report_folder}/{self.ticker}_financials.csv", index=False)
            print(f"Saved financials to {self.report_folder}/{self.ticker}_financials.csv")
        except Exception as e: print(f"Error saving financials: {e}")
    def save_price_history(self, price_history):
        try:
            price_history.to_csv(f"{self.report_folder}/{self.ticker}_prices.csv")
            print(f"Saved price history to {self.report_folder}/{self.ticker}_prices.csv")
        except Exception as e: print(f"Error saving price history: {e}")
    def save_news_sentiment(self, headlines, sentiment_results):
        try:
            df = pd.DataFrame({'Headline': headlines, 'Sentiment': [s.get('label', 'N') for s in sentiment_results], 'Confidence': [s.get('score', 0.0) for s in sentiment_results]})
            df.to_csv(f"{self.report_folder}/{self.ticker}_sentiment.csv", index=False)
            print(f"Saved sentiment data to {self.report_folder}/{self.ticker}_sentiment.csv")
        except Exception as e: print(f"Error saving sentiment data: {e}")
    def save_analysis_report(self, analysis_report):
        try:
            with open(f"{self.report_folder}/{self.ticker}_analysis_report.json", 'w') as f:
                json.dump(analysis_report, f, indent=2, default=str)
            print(f"Saved analysis report to {self.report_folder}/{self.ticker}_analysis_report.json")
        except Exception as e: print(f"Error saving analysis report: {e}")

# -------------------------
# 7. Main Orchestrator
# -------------------------
class StockAnalysisOrchestrator:
    def __init__(self, openrouter_api_key=None):
        self.openrouter_api_key = openrouter_api_key
        self.canslim_analyzer = CANSLIMAnalyzer(openrouter_api_key) if openrouter_api_key else None
        
    def analyze_stock(self, company, ticker):
        print(f"\n{'='*60}\nCOMPREHENSIVE STOCK ANALYSIS: {company} ({ticker})\n{'='*60}")
        data_agent = DataCollectorAgent(company, ticker)
        fundamental_agent = FundamentalAnalystAgent(company, data_agent)
        sentiment_agent = SentimentAnalyzerAgent()
        risk_agent = RiskManagerAgent()
        decision_agent = DecisionEngineAgent()
        persistence_manager = DataPersistenceManager(ticker)
        
        analysis_report = {'company': company, 'ticker': ticker, 'analysis_date': datetime.now().isoformat(), 'data_sources': [], 'analysis_results': {}}
        
        try:
            print("\n1. DATA COLLECTION" + "\n" + "-" * 30)
            financials = data_agent.fetch_financials(); persistence_manager.save_financials(financials); analysis_report['data_sources'].append('financials')
            price_history = data_agent.fetch_price_history(); persistence_manager.save_price_history(price_history); analysis_report['data_sources'].append('price_history')
            market_history = data_agent.fetch_market_index_history()
            headlines = data_agent.fetch_news(); analysis_report['data_sources'].append('news')
            
            print("\n2. FUNDAMENTAL ANALYSIS" + "\n" + "-" * 30)
            fundamental_analysis = fundamental_agent.comprehensive_analysis(financials)
            print(f"P/E Ratio: {fundamental_analysis.get('valuation', {}).get('pe_ratio', 'N/A')}")
            print(f"ROE: {(fundamental_analysis.get('profitability', {}).get('roe', 0) or 0)*100:.1f}%")
            print(f"Debt-to-Equity: {fundamental_analysis.get('leverage', {}).get('debt_to_equity', 'N/A')}")
            analysis_report['analysis_results']['fundamental'] = fundamental_analysis
            
            canslim_analysis = None
            if self.canslim_analyzer:
                print("\n3. CANSLIM ANALYSIS (AI-Powered)" + "\n" + "-" * 30)
                canslim_analysis = self.canslim_analyzer.analyze_canslim(company, ticker, financials, price_history, market_history, headlines)
                print(f"CANSLIM Score: {canslim_analysis.get('total_score', 'N/A')}/7 | AI Recommendation: {canslim_analysis.get('recommendation', 'N/A')}")
                for letter, data in canslim_analysis.get('canslim_analysis', {}).items(): print(f"{letter}: {'✓' if data.get('score', 0) else '✗'} - {data.get('reasoning', 'N/A')}")
                analysis_report['analysis_results']['canslim'] = canslim_analysis
            else: print("\n3. CANSLIM ANALYSIS SKIPPED (No API Key)" + "\n" + "-" * 30)
            
            print("\n4. SENTIMENT ANALYSIS" + "\n" + "-" * 30)
            sentiment_score, sentiment_results = sentiment_agent.analyze(headlines)
            persistence_manager.save_news_sentiment(headlines, sentiment_results)
            sentiment_label = 'Positive' if sentiment_score > 0.1 else 'Negative' if sentiment_score < -0.1 else 'Neutral'
            print(f"Overall Sentiment Score: {sentiment_score:.3f} ({sentiment_label})")
            analysis_report['analysis_results']['sentiment'] = {'score': sentiment_score, 'classification': sentiment_label, 'headlines_analyzed': len(headlines)}
            
            print("\n5. RISK ASSESSMENT" + "\n" + "-" * 30)
            risk_assessment = risk_agent.assess(financials, price_history)
            risk_score, risk_factors, max_risk = risk_assessment
            risk_level = 'Low' if risk_score >= 7 else 'Medium' if risk_score >= 4 else 'High'
            print(f"Risk Score: {risk_score}/{max_risk} ({risk_level})")
            for factor, level in risk_factors.items(): print(f"  {factor.replace('_', ' ').title()}: {level}")
            analysis_report['analysis_results']['risk'] = {'score': risk_score, 'max_score': max_risk, 'level': risk_level, 'factors': risk_factors}
            
            print("\n6. INVESTMENT RECOMMENDATION" + "\n" + "-" * 30)
            decision = decision_agent.decide(fundamental_analysis, sentiment_score, risk_assessment, canslim_analysis)
            print(f"RECOMMENDATION: {decision['recommendation']} | Overall Score: {decision['total_score']}/{decision['max_score']} | Confidence: {decision['confidence']*100:.1f}%")
            print(f"Reasoning: {decision['reasoning']}")
            print("\nScore Breakdown:")
            for factor, score in decision['decision_factors'].items(): print(f"  {factor.replace('_', ' ').title()}: {score:.1f}")
            analysis_report['analysis_results']['final_decision'] = decision
            
            persistence_manager.save_analysis_report(analysis_report)
            print(f"\n{'='*60}\nANALYSIS COMPLETE - All data saved for offline access\n{'='*60}")
            return analysis_report
        except Exception as e:
            print(f"Error during analysis: {e}")
            analysis_report['error'] = str(e)
            return analysis_report
@app.route('/')
def home():
    return render_template('index.html')
# -------------------------
# 8. Flask API Endpoint
# -------------------------
@app.route('/analyze', methods=['GET'])
def analyze():
    """
    Flask endpoint to trigger the stock analysis.
    Accepts 'company' and 'ticker' as query parameters.
    Example: /analyze?company=Apple Inc.&ticker=AAPL
    """
    company_name = request.args.get('company')
    ticker = request.args.get('ticker')

    if not company_name or not ticker:
        return jsonify({"error": "Please provide both 'company' and 'ticker' query parameters."}), 400

    # Initialize orchestrator
    api_key = OPENROUTER_API_KEY
    if api_key == "your_openrouter_api_key_here":
        print("WARNING: OpenRouter API key is not set. CANSLIM analysis will be skipped.")
        api_key = None # Ensure it's None if not set

    orchestrator = StockAnalysisOrchestrator(api_key)
    
    try:
        result = orchestrator.analyze_stock(company_name, ticker.upper())
        return jsonify(result)
    except Exception as e:
        print(f"A critical error occurred in the analysis orchestrator: {e}")
        return jsonify({"error": "An internal server error occurred during analysis.", "details": str(e)}), 500

if __name__ == "__main__":
    # Check for API key at startup
    if OPENROUTER_API_KEY == "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": # Example placeholder
        print("WARNING: Please replace the placeholder OpenRouter API key.")
        print("CANSLIM analysis might fail or use a generic key.")
    
    
    app.run(debug=False, port=5000)
````
